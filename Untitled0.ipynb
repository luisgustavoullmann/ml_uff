```python
def get_sinan_dataset(dis_codes: list[str], years: list[int]):
    sinan = SINAN().load()
    files = sinan.get_files(dis_code=dis_codes, year=years)
    if len(files) == 1:
        return files[0].download().to_dataframe()
    dfs = (
        f.download().to_dataframe()
        for f in files
    )
    dataset = pd.concat(dfs, ignore_index=True)
    return dataset[(dataset["SG_UF"] == "RJ") & (dataset["CLASSI_FIN"] == 10)]
dataset_sinan = get_sinan_dataset(["DENG"], [2025])
```

```python
def get_inmet():
    csvs = []
    for root, dirs, files in os.walk("sample_data/inmet"):
        for file in files:
            df = pd.read_csv(
                root + "/" + file,
                sep=";",
                decimal=",",
                encoding="latin1",
                skiprows=8,
            )
            csvs.append(df)
    return pd.concat(csvs)
inmet_dataset = get_inmet()
```

```python
def preprocess_inmet(df):
    df.columns = [c.strip() for c in df.columns]

    df["datetime"] = pd.to_datetime(
        df["Data"] + " " + df["Hora UTC"],
        errors="coerce"
    )

    df = df.replace(-9999, np.nan)

    cols = [
        "datetime",
        "TEMPERATURA DO AR - BULBO SECO, HORARIA (°C)",
        "UMIDADE RELATIVA DO AR, HORARIA (%)",
        "PRECIPITAÇÃO TOTAL, HORÁRIO (mm)",
    ]

    df = df[cols].dropna(subset=["datetime"])

    df["date"] = df["datetime"].dt.date

    df_daily = (
        df.groupby("date")
        .mean(numeric_only=True)
        .reset_index()
    )

    return df_daily


print("preprocessing inmet")
weather = preprocess_inmet(inmet_dataset)
```

```python
def preprocess_sinan(df):
    # date of symptoms
    df["date"] = pd.to_datetime(df["DT_SIN_PRI"], errors="coerce").dt.date

    zika_daily = (
        df.groupby("date")
        .size()
        .reset_index(name="cases")
    )

    return zika_daily


print("preprocessing sinan")
dengue = preprocess_sinan(dataset_sinan)
```

```python
df = pd.merge(weather, dengue, on="date", how="left")
df["cases"] = df["cases"].fillna(0)

df["date"] = pd.to_datetime(df["date"])

df["month"] = df["date"].dt.month
df["weekofyear"] = df["date"].dt.isocalendar().week.astype(int)

for lag in [7, 14, 21]:
    df[f"cases_lag_{lag}"] = df["cases"].shift(lag)

df["temp_7d"] = df[
    "TEMPERATURA DO AR - BULBO SECO, HORARIA (°C)"
].rolling(7).mean()

df["rain_7d"] = df[
    "PRECIPITAÇÃO TOTAL, HORÁRIO (mm)"
].rolling(7).sum()
df = df.dropna()
```

```python

features = df.drop(columns=["cases", "date", "temp_7d", "rain_7d"])
target = df["cases"]

model = xgb.XGBRegressor(
    objective="count:poisson",
    n_estimators=400,
    max_depth=5,
    learning_rate=0.05,
    subsample=0.8,
    colsample_bytree=0.8
)

model.fit(features, target)
```
